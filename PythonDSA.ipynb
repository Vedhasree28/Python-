```python
# Stack Implementation using ADT(Abstract Data Type-classes & objects)
class Stack:
  capacity=5
  def __init__(self):
    self.stack=[]
  def push(self,element):
    if len(self.stack)>=self.capacity:
      return("Stack Overflow")
    else:
      return self.stack.append(element)
  def pop(self):
    if self.stack:
      return(f"Popped element: {self.stack.pop()}")
    else:
      return("Stack Underflow")
  def peek(self):
    if self.stack:
      return(f"Topmost element: {self.stack[-1]}")
    else:
      return("Stack Underflow")
  def isempty(self):
    if len(self.stack)==0:
      return("Stack is Empty")
    else:
      return self.stack
  def display(self):
    return(f"Stack:{self.stack}")
if __name__=="__main__":
  s=Stack()

  print(s.isempty())
  s.push(10)
  s.push(20)
  s.push(30)
  print(s.display())
  print(s.pop())
  print(s.peek())
  print(s.display())
```

```python
# Reverse of a str using functions
def reverseword(word):
  stack=[]
  rev=[]
  for ch in word:
    stack.append(ch)           #Pushing Individual characters
  while stack:
    rev.append(stack.pop())
  return rev
word=input("Enter any word:")
print(reverseword(word))
```

```python
# Reverse of a str using functions
def reverseword(word:str)->str:     #Type printing
  stack=[]
  rev=[]
  for ch in word:
    stack.append(ch)         #Pushing Individual characters
  while stack:
    rev.append(stack.pop())
  return("".join(rev))
word=input("Enter any word:")
print(reverseword(word))
```

```python
# Removing Adjacent Duplicates
word="abbaca"
stack=[]
for ch in word:
  if stack and ch==stack[-1]:
    stack.pop()
  else:
    stack.append(ch)

print("".join(stack))
```

```python
# Removing Adjacent Duplicates Using Functions
def rem_duplicates(word:str)->str:
  stack=[]
  for ch in word:
    if stack and ch==stack[-1]:
      stack.pop()
    else:
      stack.append(ch)
  return("".join(stack))
word="abbaca"
print(rem_duplicates(word))
```

```python
# Balanced Paranthesis using stack
s="([]{})"
stack=[]
pairs={
    "]":"[",
    "}":"{",
    ")":"("
}
balanced=True
for ch in s:
  if ch in "([{":               # if ch in pairs.values:
    stack.append(ch)
  else:
    if not stack or stack.pop() !=pairs[ch]:
      balanced=False
      break
if balanced and not stack:
  print("Balanced")
else:
  print("Not Balanced")
```
