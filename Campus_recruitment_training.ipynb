```python
# Using for loop
a=int(input("Enter first term:"))
terms=int(input("No.of terms:"))
for i in range(1,terms+1):
  print(a,end=" ")
  a=a*i+i
```

```python
# Using while loop
a=int(input("Enter first term:"))
terms=int(input("No.of terms:"))
i=1
while(a>0):
  print(a,end=" ")
  a=a*i+i
  i=i+1
  if i==terms+1:
    break
```

## **Four ways to write a reverse of string**

**1.Using slicing**

```python
word=input("enter any string:")
print("Reverse of a String:",word[::-1])      #OR
print("Reverse of a String:",word[len(word):-(len(word)+1):-1])   #OR
print("Reverse of a String:",word[len(word):None:-1])
```

**2.Using loop**

```python
word=input("enter any string:")
rev=""
for ch in word:
    rev=ch+rev
print("Reverse of a String:",rev)
```

**3.Using built-in Methods**

```python
word=input("enter any string:")
print("Reverse of a String:","".join(reversed(word)))
```

**4.Using Recursive functions**

```python
word=input("Enter any str:")
def str_rev(word):
  rev=""
  for ch in word:
    rev=ch+rev
  return rev
print(str_rev(word))
```

```python
def str_rev(word):
  if word=="":
    return word
  return word[-1]+str_rev(word[-(len(word)+1):-1])
word=input("Enter any str:")
print(str_rev(word))
```

## **Three Ways to Iterate over a String**

**Using Membership Operator**

```python
word="Placement"
for ch in word:
  print(ch,end="")
print()
```

**Using range method**

```python
word="Placement"
for i in range(len(word)):
  print(word[i],end="")
print()
```

**Using Enumerate**

```python
word="Placement"
for i in enumerate(word):
  print(i,end="")
```

```python
word="Placement"
for v,i in enumerate(word):
  print(v,i)
```

# **Word Count**

```python
# Using flag
s="I love Python"
count=0
word=False
for ch in s:
  if ch != " ":
    if word==False:
      count+=1
      word=True
  else:
    word=False

print("Word count:",count)
```

**WAP to find the duplicated elements in a string**

```python
# Printing all duplicated character using count() method
word=input("Enter any string:")
dup=""
for ch in word:
  if word.count(ch)>1 and ch not in dup:
    dup+=ch
print(dup)
```

```python
# Print duplicated character without using count()
word=input("Enter any string:")
unique=""
dup=""
for ch in word:
  if ch not in dup and ch in unique:
    dup+=ch
  else:
    unique+=ch
print(dup)
```

```python
# Print duplicated character without using count()
word=input("Enter any string:")
unique=""
dup=""
for ch in range(len(word)):
  if word[ch] not in unique:
    unique+=word[ch]
  elif word[ch] not in dup:
    dup+=word[ch]
print(dup)
```
