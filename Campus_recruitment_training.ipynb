**Reasoning problem: 3,4,10,33,136,?**



```python
# Using for loop
a=int(input("Enter first term:"))
terms=int(input("No.of terms:"))
for i in range(1,terms+1):
  print(a,end=" ")
  a=a*i+i
```

```python
# Using while loop
a=int(input("Enter first term:"))
terms=int(input("No.of terms:"))
i=1
while(a>0):
  print(a,end=" ")
  a=a*i+i
  i=i+1
  if i==terms+1:
    break
```
