```python
# sum of elements in lst using while loop
nums=[10,12,5,2,20]
count=0
n=0
while n<len(nums):
  count+=nums[n]
  n+=1
print(count)
```

```python
# method 2 : Using for loop
nums=[10,12,5,2,20]
count=0
for i in nums:
  count+=i
print(count)
```

```python
# Multiplication of elements in a list
nums=[10,12,5,2,20]
count=1
for i in nums:
  count*=i
print(count)
```

```python
# Subtraction of elements in a list
nums=[10,12,5,2,20]
count=nums[0]
for i in range(1,len(nums)):
  count-=nums[i]
print(count)
```
