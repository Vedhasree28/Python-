**6. Write a program to print the first N terms of the Fibonacci series.**

```python
n=int(input("enter:"))
a,b=0,1
for i in range(n):
  print(a,end=" ")
  a,b=b,a+b
```

**7. Write a program to reverse a given integer number (e.g. 123 -> 321).**

```python
n=int(input("Enter:"))
rev=0
while n>0:
  rem=n%10
  rev=rev*10+rem
  n=n//10
print(rev)
```

**8. Write a program to check whether a given number is a palindrome.**

```python
n=int(input("Enter:"))
rev=0
temp=n
while n>0:
  rem=n%10
  rev=rev*10+rem
  n=n//10
if temp==rev:
  print("Palindrome")
else:
  print("Not palindrome")
```

**9. Write a program to find the sum of digits of a given number.**

```python
n=int(input("Enter"))
sum=0
while n>0:
  rem=n%10
  sum+=rem
  n=n//10
print(sum)
```

**10. Write a program to check whether a given year is a leap year.**

```python
year=int(input("enter any year:"))
if (year%400 == 0) | (year%4 ==0) & (year%100!= 0):
  print("Leap year")
else:
  print("not leap year")
```
