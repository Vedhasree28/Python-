# if-elif-else

```python
a=int(input('enter a num:'))
if a>0:
    print(a,'is a pos number')
    print('iam in if block')
elif a==0:
    print(a,'is zero')
    print('iam in elif block')
else:
    print(a,'is a neg num')
    print('iam in else block')
```

# Iterative statements

### while loop:repeatedly executes a target statement until the given condition is true

```python
x=0
while x<10:
    print(x)
    x+=1
```

### for loop:it has the ability to iterate over items of any sequence such as list,strings,tuples,dict etc..

```python
values=range(15,30)
for i in values:
    print(i)
```

```python
dict1={1:1,2:4,3:9,4:16}
for i,j in dict1.items():
    print('key:',i,'value:',j)
```

# Transfer statements

### Break :it brings control out of the loop

```python
for i in range(1,10):
    if i==6:
        break
    print(i)
```

### continue:it returns control to the begining of the loop

```python
for i in range(1,10):
    if i==6:
        continue
    print(i)
```
