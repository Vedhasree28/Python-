**11. Write a program to count the number of vowels in a given string.**

```python
n=input("Any string:")
count=0
for ch in n:
  if ch in "aeiou":
    count+=1
print(count)
```

**12. Write a program to reverse a given string without using slicing.**

```python
n=input("enter:")
rev=""
for i in n:
  rev=i+rev
print(rev)
```

**13. Write a program to check whether a given string is a palindrome.**

```python
s=input("Enter:")
if s==s[::-1]:
  print("Palindrome")
else:
  print("Not palindrome")
```

**14. Write a program to count the number of words in a given sentence.**

```python
s=input("Enter any sentence")
count=0
word=False
for ch in s:
  if ch != " ":
    if word == False:
      count+=1
      word=True
  else:
    word=False
print(count)
```

**15. Write a program to convert a string to uppercase and lowercase without using built-in case methods.**

```python
#lower to upper
n=input("Enter:")
def upper(n):
  result=""
  for char in n:
    if 'a' <= char <= 'z':
      result+=chr(ord(char)-32)
    else:
      result+=char
  return result
result=upper(n)
print(result)
```

```python
#upper to lower
n=input("Enter")
def lower(n):
  result=""
  for ch in n:
    if "A" <= ch <= "Z":
      result+=chr(ord(ch)+32)
    else:
      result+=ch
  return result
result=lower(n)
print(result)
```
