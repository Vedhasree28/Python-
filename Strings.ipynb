**11. Write a program to count the number of vowels in a given string.**

```python
n=input("Any string:")
count=0
for ch in n:
  if ch in "aeiou":
    count+=1
print(count)
```

**12. Write a program to reverse a given string without using slicing.**

```python
n=input("enter:")
rev=""
for i in n:
  rev=i+rev
print(rev)
```
