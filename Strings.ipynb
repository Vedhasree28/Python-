**11. Write a program to count the number of vowels in a given string.**

```python
n=input("Any string:")
count=0
for ch in n:
  if ch in "aeiou":
    count+=1
print(count)
```

**12. Write a program to reverse a given string without using slicing.**

```python
n=input("enter:")
rev=""
for i in n:
  rev=i+rev
print(rev)
```

```python
n=input("enter:")
for i in range(len(n)-1,-1,-1):
  print(n[i],end="")
```

**13. Write a program to check whether a given string is a palindrome.**

```python
s=input("Enter:")
if s==s[::-1]:
  print("Palindrome")
else:
  print("Not palindrome")
```

**14. Write a program to count the number of words in a given sentence.**

```python
s=input("Enter any sentence")
count=0
word=False
for ch in s:
  if ch != " ":
    if word == False:
      count+=1
      word=True
  else:
    word=False
print(count)
```
