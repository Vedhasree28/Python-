**21. Write a program to find the largest and smallest elements in a list.**

```python
a=[1,4,3,3]
largest=a[0]
smallest=a[0]
for i in range(len(a)):
  if a[i] > largest:
    largest=a[i]
  elif a[i] < smallest:
    smallest=a[i]
print("Largest:",largest)
print("Smallest:",smallest)
```


**22. Write a program to calculate the sum and average of all elements in a list.**

```python
lst=[1,4,3,2]
print("Sum of all elements is:",sum(lst))
print("Average of all elements is:",sum(lst)/len(lst))
```

```python
lst=[1,4,3,2]
sum=0
for i in range(len(lst)):
  sum=sum+lst[i]
  avg=sum/len(lst)
print("Sum:",sum)
print("avg:",avg)
```

```python
lst=[1,4,3,2]
sum=0
for i in lst:
  sum=sum+i
  avg=sum/len(lst)
print("Sum:",sum)
print("avg:",avg)
```

**23. Write a program to count the number of even and odd numbers in a list.**

```python
lst=[2,3,5,4,8,0]
even_count=0
odd_count=0
for i in lst:
  if i%2==0:
    even_count+=1
  else:
    odd_count+=1
print("Even numbers count:",even_count)
print("odd numbers count:",odd_count)
```

**26. Write a program to find the second largest element in a list.**

```python
lst=[2,4,3,6,5]
lst.sort()
print("second largest element:",lst[-2])
```

```python
lst=[2,4,3,6,5]
highest = float('-inf')  #works for even negative numbers in lst
second = float('-inf')
for i in lst:
  if i>highest:
    second=highest
    highest=i
  elif highest>i>second:
    second=i
print("Second largest element:",second)
```

**27. Write a program to reverse a list without using the reverse() method.**
