**21. Write a program to find the largest and smallest elements in a list.**

```python
a=[1,4,3,3]
largest=a[0]
smallest=a[0]
for i in range(len(a)):
  if a[i] > largest:
    largest=a[i]
  elif a[i] < smallest:
    smallest=a[i]
print("Largest:",largest)
print("Smallest:",smallest)
```
