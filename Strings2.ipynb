**16. Write a program to count the frequency of each character in a string.**

```python
s=input("enter:")
freq={}
for ch in s:
  if ch in freq:
    freq[ch]+=1
  else:
    freq[ch]=1
print(freq)
```

**17. Write a program to remove all duplicate characters from a string.**

```python
s=input("enter:")
res=""
for ch in s:
  if ch not in res:
    res+=ch
print(res)
```

**18. Write a program to check whether two strings are anagrams of each other.**

```python
s1="listen"
s2="silent"
if sorted(s1) == sorted(s2):
  print("Anagrams")
else:
  print("not Anagrams")
```
