**Problem statement:
Riya travels on the city metro with a smart card that starts with a balance of B rupees. She has planned N trips, numbered 1 to N in the order she will take them. Trip i has a fare of F_i rupees.
The card follows these rules, applied to each trip in order:
1. Loyalty rule. If the trip number is divisible by K, the trip is free. No fare is deducted and no top-up happens for that trip.
2. Auto top-up rule. For any other trip, if the balance is less than the fare, the card buys top-up units before the trip. Each unit adds exactly T rupees. The card buys the smallest number of units that makes the balance at least equal to the fare.
3. The fare is then deducted from the balance. Find the total number of top-up units bought across all trips and the balance left on the card after the last trip.
Input format
Line 1: four integers N, B, T and K.
Line 2: N integers F_1, F_2, …, F_N. Output format
One line with two integers separated by a space: the total number of top-up units bought, and the final balance.
Constraints 1 ≤ N ≤ 2 × 10^5 0 ≤ B ≤ 10^9 1 ≤ T ≤ 10^9 1 ≤ K ≤ 10^9 1 ≤ F_i ≤ 10^12
Sample input
6 50 100 4
30 45 320 500 60 15
Sample output 5 80 **

```python
N,B,T,K=map(int,input().split())
f_N=list(map(int,input().split()))
count=0
for i in range(N):
    if (i+1)%K==0:
        continue
    elif B>=f_N[i]:
        B=B-f_N[i]
    else:
        while B<f_N[i]:
            B=B+T
            count=count+1
        B=B-f_N[i]
print(count,B)
```

```python
import sys
def main():
  sys.stdin = open("input.txt", "r")
  data=sys.stdin.buffer.read().split()
  N,B,T,K=int(data[0]),int(data[1]),int(data[2]),int(data[3])
  units=0
  for i in range(1,N+1):
    if i%K==0:
      continue
    fares=int(data[3+i])
    if B<fares:
      need=(fares-B+T-1)//T
      units+=need
      B+=need*T
    B-=fares
  print(units,B)
main()
```
