**Problem statement:
Riya travels on the city metro with a smart card that starts with a balance of B rupees. She has planned N trips, numbered 1 to N in the order she will take them. Trip i has a fare of F_i rupees.
The card follows these rules, applied to each trip in order:
1. Loyalty rule. If the trip number is divisible by K, the trip is free. No fare is deducted and no top-up happens for that trip.
2. Auto top-up rule. For any other trip, if the balance is less than the fare, the card buys top-up units before the trip. Each unit adds exactly T rupees. The card buys the smallest number of units that makes the balance at least equal to the fare.
3. The fare is then deducted from the balance. Find the total number of top-up units bought across all trips and the balance left on the card after the last trip.
Input format
Line 1: four integers N, B, T and K.
Line 2: N integers F_1, F_2, …, F_N. Output format
One line with two integers separated by a space: the total number of top-up units bought, and the final balance.
Constraints 1 ≤ N ≤ 2 × 10^5 0 ≤ B ≤ 10^9 1 ≤ T ≤ 10^9 1 ≤ K ≤ 10^9 1 ≤ F_i ≤ 10^12
Sample input
6 50 100 4
30 45 320 500 60 15
Sample output 5 80 **

```python
N,B,T,K=map(int,input().split())
f_N=list(map(int,input().split()))
count=0
for i in range(N):
    if (i+1)%K==0:
        continue
    elif B>=f_N[i]:
        B=B-f_N[i]
    else:
        while B<f_N[i]:
            B=B+T
            count=count+1
        B=B-f_N[i]
print(count,B)
```

```python
%%writefile input.txt
6 50 100 4
30 45 320 500 60 15
```

```python
import sys
def main():
  sys.stdin = open("input.txt", "r")
  data=sys.stdin.buffer.read().split()
  N,B,T,K=int(data[0]),int(data[1]),int(data[2]),int(data[3])
  units=0
  for i in range(1,N+1):
    if i%K==0:
      continue
    fares=int(data[3+i])
    if B<fares:
      need=(fares-B+T-1)//T
      units+=need
      B+=need*T
    B-=fares
  print(units,B)
main()
```

# **Question 2: Courier Sorting Hub**
**A courier hub receives N parcels. Each parcel carries a code made of English letters, digits and hyphens. The label printer at the source warehouse is faulty: it may print the characters of a code in any order, may switch letters between upper and lower case, and may insert or drop hyphens. It never adds, removes or changes a letter or a digit.
Two parcels therefore belong to the same destination bin if, after removing all hyphens and ignoring letter case, one code can be rearranged to give the other. Every character counts with its multiplicity, so AAB and ABB belong to different bins. Find how many bins the hub needs and how many parcels the fullest bin holds.
Input format
Line 1: an integer N.
Next N lines: one parcel code per line.
Output format
Line 1: the number of bins.
Line 2: the number of parcels in the fullest bin.
Constraints
1 ≤ N ≤ 10^5
Each code has 1 to 30 characters, drawn from AZ , a–z, 0–9 and -.
Each code contains at least one letter or digit, and contains no spaces.**

```python
%%writefile input2.txt
7
AB-12
b1a2
21BA
XY-9
9yx
Q7
AB-1
```

```python
import sys
def main():
  sys.stdin=open('input2.txt','r')
  data=sys.stdin.read().split()
  N=int(data[0])
  bin={}
  for code in data[1:N+1]:
    key=''.join(sorted(code.lower().replace('-','')))
    bin[key]=bin.get(key,0)+1
  print(len(bin))
  print(max(bin.values()))
main()
```
